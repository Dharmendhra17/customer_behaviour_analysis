```python
# create a age group 
labels = ['Young Adult','Adult','Middle-age','Senior']
df['age_group']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','age_group']].head()
```

```python
#create a colums purchase a frequency days 
frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequnency_days'] = df['frequencyofpurchases'].map(frequency_mapping)
```
